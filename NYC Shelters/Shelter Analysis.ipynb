```python
import pandas as pd
```

```python
df = pd.read_csv("Shelters.csv")
```

```python
cols_to_drop = [
    'Adult Family Commercial Hotel',
    'Adult Shelter Commercial Hotel',
    'Family Cluster',
    'Family with Children Commercial Hotel'
]

df = df.drop(columns=cols_to_drop)
```

```python
print(df.head())
```

```python
#Add "Year" column
df['Report Date'] = pd.to_datetime(df['Report Date'], format='%m/%d/%Y')
df['Year'] = df['Report Date'].dt.year
```

```python
# Get unique combos per year as sets, see what's changed/removed/added year after year
yearly = df.groupby('Year').apply(
    lambda x: set(zip(x['Community Districts'], x['Census Type']))
).to_dict()

years = sorted(yearly.keys())
for i in range(1, len(years)):
    prev, curr = years[i-1], years[i]
    added = yearly[curr] - yearly[prev]
    removed = yearly[prev] - yearly[curr]
    print(f"\n{prev} → {curr}")
    print(f"  Added:   {added}")
    print(f"  Removed: {removed}")
```
